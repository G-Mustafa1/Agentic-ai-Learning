```python
sum1 = int(input("Addition value 1"))
sum2 = int(input("Addition value 2"))
adding = sum1 + sum2
print(f'your value 1 is {sum1} value 2 is {sum2} adding value is  {adding}')
```

```python
sum3 = int(input("Multiply value 1"))
sum4 = int(input("Multiply value 2"))
multiply = sum3 * sum4
print(f'your value 1 is {sum3} value 2 is {sum4} Multiply value is {multiply}')
```

```python
sum5 = int(input("Devided value 1"))
sum6 = int(input("Devided value 2"))
devided = sum5 / sum6
print(f'your value 1 is {sum5} value 2 is {sum6} Devided value is {devided}')
```

```python
sum7 = int(input("Subtract value 1"))
sum8 = int(input("Subtract value 2"))
subtract = sum7 - sum8
print(f'your value 1 is {sum7} value 2 is {sum8} Subtract value is {subtract}')
```
